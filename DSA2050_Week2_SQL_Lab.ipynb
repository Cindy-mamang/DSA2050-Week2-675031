```python
import sqlite3
import pandas as pd

# 1. Connect to your database
conn = sqlite3.connect('your_database.db')

# 2. Load your original CSV files into pandas DataFrames
# (Make sure 'orders.csv' and 'customers.csv' are in the same folder as your script)
df_orders = pd.read_csv('orders.csv')
df_customers = pd.read_csv('customers.csv')

# 3. Write/save those DataFrames into the SQLite database as tables
df_orders.to_sql('orders', conn, index=False, if_exists='replace')
df_customers.to_sql('customers', conn, index=False, if_exists='replace')

print("Tables successfully loaded into the database!")

# 4. Now run your JOIN query safely
query = """ 
SELECT 
    o.OrderID, 
    o.CustomerID, 
    c.CustomerName, 
    c.Segment, 
    o.OrderAmount 
FROM orders o 
LEFT JOIN customers c 
ON o.CustomerID = c.CustomerID; 
""" 

# 5. Execute and view results
joined = pd.read_sql_query(query, conn)
print(joined.head())

# 6. Sanity check lengths
print("Original orders count:", len(pd.read_sql_query("SELECT * FROM orders", conn)))
print("Joined data count:", len(joined))

# 7. Close connection
conn.close()
```

```python
# Q8. Are the row counts equal? If not, explain exactly what caused the difference. 
# Yes, the row counts are equal. A LEFT JOIN preserves every single row 
# from the left table (orders). Because no order rows are filtered out or 
# dropped during the merge, the total row count of the joined table matches 
# the original orders table.
```

```python
# Q9. Locate the unmatched customer record in the joined output. 
# What values appear for CustomerName and Segment?
# The values that appear for the unmatched customer record are None 
# (or NaN / NULL) for both CustomerName and Segment. When an order contains 
# a CustomerID with no matching entry in the customers table, the LEFT JOIN 
# retains the order details but assigns null placeholders for the missing attributes.
```

```python
# 9. Part H – Reconcile the Totals 
# Q10. Are the two sales totals equal?
# Yes, the two sales totals are equal. Summing the OrderAmount column 
# before and after the JOIN yields the exact same total revenue figure.


# Q11. If the total changed, which customer caused the duplication?
# Since a LEFT JOIN with a 1-to-1 or left-dominant relationship does not 
# duplicate rows, the total did not change. Therefore, no customer caused 
# a duplication, and the sales totals remain identical.


# Q12. Why is total reconciliation an important validation step after a JOIN?
# Total reconciliation is a critical validation step because accidental 
# many-to-many relationships during a join can silently duplicate rows and 
# inflate financial metrics. Verifying that totals match ensures data integrity 
# and prevents reporting distorted revenue or volume figures.
```

```python
# Part I – Correct the Duplicate-Key Problem 
# Create a customer table with one record per CustomerID, then repeat the JOIN.

import sqlite3
import pandas as pd

# 1. Connect to your database
conn = sqlite3.connect('your_database.db')

# 2. Load your original CSV files into pandas DataFrames
orders = pd.read_csv('orders.csv')
customers = pd.read_csv('customers.csv')

# 3. Save the orders table into SQLite so the SQL query can find it
orders.to_sql("orders", conn, if_exists="replace", index=False)

# 4. Clean duplicate customer records and save to SQLite
customers_clean = customers.drop_duplicates(subset="CustomerID") 
customers_clean.to_sql( 
    "customers_clean", 
    conn, 
    if_exists="replace", 
    index=False 
) 

# 5. Define the SQL LEFT JOIN query
query = """ 
SELECT 
    o.OrderID, 
    o.CustomerID, 
    c.CustomerName, 
    c.Segment, 
    o.OrderAmount 
FROM orders o 
LEFT JOIN customers_clean c 
    ON o.CustomerID = c.CustomerID; 
""" 

# 6. Execute query and display results
clean_joined = pd.read_sql_query(query, conn) 
print(clean_joined.head())

# 7. Print row counts and sales totals for reconciliation
print("\nRow counts (orders, clean_joined):", len(orders), len(clean_joined))
print("Sales totals (orders sum, clean_joined sum):", orders["OrderAmount"].sum(), clean_joined["OrderAmount"].sum())

# Close connection
conn.close()
```

```python
# Q13. Has the corrected JOIN preserved both the row count and the total sales value? Explain.
# Yes, the corrected JOIN has successfully preserved both the row count and 
# the total sales value. 

# Explanation: By dropping duplicate CustomerID records from the customers table 
# before performing the LEFT JOIN, we eliminated accidental many-to-many 
# relationships or row duplication. As a result, each order matches with exactly 
# one unique customer record, keeping the total row count equal to the original 
# orders table and maintaining an identical total sales sum without any inflation.
```

```python
# 11. P
# art J – Customer Segment KPIs
import sqlite3
import pandas as pd

# 1. Connect to your database
conn = sqlite3.connect('your_database.db')

# 2. Define the SQL query using GROUP BY and aggregate functions
query = """ 
SELECT 
    c.Segment, 
    COUNT(o.OrderID) AS OrderCount, 
    SUM(o.OrderAmount) AS TotalSales, 
    AVG(o.OrderAmount) AS AverageOrderValue 
FROM orders o 
LEFT JOIN customers_clean c 
    ON o.CustomerID = c.CustomerID 
GROUP BY c.Segment; 
""" 

# 3. Execute the query and load into a DataFrame
segment_kpis = pd.read_sql_query(query, conn) 

# 4. Display the results in VS Code
print("--- Customer Segment KPIs ---")
print(segment_kpis)

# Close connection
conn.close()
```

```python
# 12. Part K – Merge the JSON Region Lookup 
# 1. Load the region names from the JSON file
regions_df = pd.read_json("regions.json") 

# 2. Merge the cleaned customer table with the regions DataFrame
customers_regions = customers_clean.merge( 
    regions_df, 
    on="RegionCode", 
    how="left" 
) 

# Preview the merged data
print(customers_regions.head())
```

```python
# Q15. Confirm that each customer now has a readable RegionName rather 
# than only a RegionCode.
# Yes, confirmed. By performing a left merge on the shared RegionCode column, 
# the resulting customers_regions DataFrame successfully maps each customer's 
# RegionCode to its corresponding full descriptive RegionName from the JSON source.


# Q16. If a RegionCode did not exist in regions.json, what would a left merge 
# produce for RegionName?
# If a RegionCode did not exist in regions.json, the left merge would produce 
# null values (NaN or None) for the RegionName column for any customers 
# associated with that unmatched code, while still retaining all customer records.
```

```python
# 13. Final Challenge – Regional Sales Analysis 
import sqlite3
import pandas as pd

# 1. Connect to your database
conn = sqlite3.connect('your_database.db')

# 2. Save the fully merged customers_regions table to SQLite so SQL can query it
customers_regions.to_sql("customers_regions", conn, if_exists="replace", index=False)

# 3. Define the SQL query to aggregate order count and total sales by region
query = """ 
SELECT 
    cr.RegionName, 
    COUNT(o.OrderID) AS NumberOfOrders, 
    SUM(o.OrderAmount) AS TotalSales 
FROM orders o 
LEFT JOIN customers_regions cr 
    ON o.CustomerID = cr.CustomerID 
GROUP BY cr.RegionName; 
""" 

# 4. Execute the query and load into a DataFrame
region_summary = pd.read_sql_query(query, conn) 

# 5. Display the final table in VS Code
print("--- Orders and Total Sales by Region ---")
print(region_summary)

# Close connection
conn.close()
```
